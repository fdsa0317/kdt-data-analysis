### shape
    - numpy에서는 해당 array의 크기를 알 수 있다. shape를 확인함으로서, 몇 개의 데이터가 있는지 몇 차원으로 존재하는지 확인할 수 있다.
    - arr.shape의 결과는 튜플로 (행, 열)로 행만 존재하면 1차원, 행렬이 존재하면 2차원으로 차수를 알 수 있다. 또한 행*렬을 곱하면 크기도 확인할 수 있다.

```python
print(arr3)
print(arr3.shape)

# (10, )
# 1차원, 행의 10개 데이터

# 1차원 벡터
```

## 브로드캐스팅
- 다른 크기의 배열들 간에도 자동으로 연산해주는 Numpy의 강력한 기능

## 브로드캐스팅의 특징
    - 기본적으로는 shape와 같은 두 ndarray에 대한 연산은 각 원소별로 진행한다
    - 다른 shape를 갖는 array 간의 연산의 경우 브로드 캐스팅(shape를 맞춤) 후 진행한다

<img width=500 src='https://www.astroml.org/_images/fig_broadcast_visual_1.png' />

```python
arr1 = np.arange(12).reshape(4, 3)

print(arr1)
print(arr1.shape)
```

```python
arr2 = np.arange(3)

print(arr2)
print(arr2.shape)
```

```python
print(arr1 + arr2)
```

## 브로드 캐스팅의 규칙
1. 뒷 차원부터 비교한다.

3. 차원 중 값이 1이 존재해야 한다.

[브로드캐스팅 공식문서](https://numpy.org/doc/stable/user/basics.broadcasting.html)

```python
"""
    차원에 1이 존재하는 경우 확장이 가능하다.
    6 x 1 x 5
        7 x 3

    6 x 7 x 5
"""

None
```

```python
arr1 = np.arange(100, 104)
print(arr1)
```

```python
# (4, 3)
arr2 = np.arange(12).reshape(4, 3)
print(arr2)
```

```python
arr1 + arr2
```

<img src="https://numpy.org/doc/stable/_images/broadcasting_3.png" />

```python
# (2, 3), (3, 2)
arr3 = np.arange(6).reshape(2, 3)
arr4 = np.arange(100, 106).reshape(3, 2)

print(arr3)
print(arr4)
```

```python
arr3.T + arr4
```

## boolean selection, boolean indexing
- array 인덱싱 시, bool 타입의 조건에 맞는 list를 전달하여 조건식을 만족하는 True의 결과물만 필터링한다.

```python
arr4 = np.random.randint(1, 100, size=20)
print(arr4)
```

```python
even_mask = arr4 % 2 == 0
arr4[even_mask]
```

```python
is_bound = arr4 > 30
arr4[is_bound]
```

## 다중 조건식
- &: and 조건식
- |: or 조건식

```python
# 홀수이면서 50보다 작은 값s
# and: 
np.random.seed(2026)
arr5 = np.random.randint(100, size=15)
arr5
```

```python
is_bound = (arr5 % 2 == 1) & (arr5 < 50) 
arr5[is_bound]
```

```python
arr5[is_bound].astype(str)
```
