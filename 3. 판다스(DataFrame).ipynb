## 파생 변수 만들기 연습

```python
np.nan == np.nan
```

```python
# 만약 NaN이라면: 알수없음
# 19세 이상: 성인
# 19세 미만: 아이
# if ~ elif ~ else
def get_adult(age):
    if np.isnan(age):
        return "알 수 없음"

    return "성인" if age >= 19 else "아이"

get_adult(19)
```

```python
# .apply(): 앞의 시리즈의 값을 순차적으로 받아서 결과를 바꾸고 새로운 결과를 반환하는 함수
titanic_df["IsAdult"] = titanic_df["Age"].apply(get_adult)
titanic_df
```

```python
# RipOff
# Fare > 300: 바가지 요금
# Fare <= 300: 정상 요금
```

```python
titanic_df["RipOff"] = titanic_df["Fare"].apply(lambda fare: "바가지" if fare > 300 else "정상")
titanic_df
```

```python
titanic_df[titanic_df["Fare"] > 300]
```

```python
# 파생변수 IsAdult, RipOff 두 개를 모두 삭제
```

```python
titanic_df.drop(["IsAdult", "RipOff"], axis=1, inplace=True)
```
